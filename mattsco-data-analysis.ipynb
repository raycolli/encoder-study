```python
### Look into Mattsco dataset for next trial iteration

import pandas as pd

df = pd.read_excel('data\Mattsco Product Parsed.Marty.12.19.24.xlsx')

# lets look at top 30 items per column to get a better feel for the distribution.
import matplotlib.pyplot as plt
import seaborn as sns

# Loop through categorical columns
for column in df.select_dtypes(include=['object', 'category']).columns:
    # Get the top 10 most recurring items
    top_items = df[column].value_counts().nlargest(30)
    
    # Calculate total count of items in the column
    total_count = df[column].count()
    
    # Create a new figure for each plot
    plt.figure(figsize=(10, 6))
    
    # Plot the bar chart
    ax = sns.barplot(x=top_items.index, y=top_items.values)

    print('top items: ', top_items.index)
    
    # Add count on top of each bar
    for p in ax.patches:
        ax.annotate(f'{int(p.get_height())}', 
                    (p.get_x() + p.get_width() / 2., p.get_height()), 
                    ha='center', va='bottom')

    # Set the title and labels
    plt.title(f"Top 30 items in '{column}' (Total Count: {total_count})")
    plt.xlabel(column)
    plt.ylabel('Frequency')
    plt.xticks(rotation=90)  # Rotate x labels for better readability
    
    # Show the plot
    plt.tight_layout()
    plt.show()
```
